```python
 import pandas as pd
 import numpy as np
 from sklearn.linear_model import LinearRegression
 import matplotlib.pyplot as plt
 import pickle
```

```python
saldf=pd.read_csv('/content/drive/MyDrive/parcel_cost.csv')
```

```python
 saldf.head()
```

```python
saldf.info()
```

```python
saldf.isnull().sum()
```

```python
inp=saldf[['weight_kg']]
out=saldf['cost_rupees']
```

```python
LR=LinearRegression()
```

```python
LR.fit(inp,out)
```

```python
with open("parcel.pkl", "wb") as f:
    pickle.dump(LR, f)

print("Model saved successfully")
```
